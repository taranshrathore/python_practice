```python
#inheritance

class Animal:
    def __init__(self,name):
        self.name = name

    def speak(self):
        return  f"{self.name} says woof!"


class Dog(Animal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed = breed

    def fetch(self):
        return f"{self.name}  is of the breed {self.breed}"

d1 = Dog("sherry", "Rottweiler")


print(d1.name)
print(d1.fetch())
print(d1.speak())
```

```python
class Vehicle:  #made a parent class vehicle
    def __init__(self, brand, speed):  #initial function dunder method init whic python calls itself, self is a must rest two attributes can be specific
        self.speed = speed #always has to make sure atrributes are called with self
        self.brand = brand

    def is_fast(self):  #funtion to chekc if car is ffast enough
        return self.speed > 120  #speed would be passed and check should be more than 120

    def __str__(self):   #tells python what text represents this object, could use print but it just displays shit on screen
        return f"Brand: {self.brand} | Speed: {self.speed}"


class ElectricVehicle(Vehicle):
    def __init__(self, brand, speed, battery_range):
        super().__init__(brand, speed) #super is used to inherit methoda and attributes from other classes or parent class
        self.battery_range = battery_range

    def show_range(self):
        return f"{self.brand} has a range of {self.battery_range} km"

    def __str__(self):
        return super().__str__() + f" | Range: {self.battery_range} km"


v1 = Vehicle("Mahindra XUV", 200)
print(v1)
print(v1.is_fast())

e1 = ElectricVehicle("Skoda Taigun", 180, 550)
print(e1)
print(e1.is_fast())
print(e1.show_range())
```

```python
# Iterators and Generators 

def even_numbers(n):
    for i in range(0,n+1):
        if i % 2 ==  0:
            yield i

for num in even_numbers(10):
    print(num)
```

```python
def squares(n):
    for square in range(1, n + 1):
        yield square ** 2

gen = squares(3)
print(next(gen))     # 1
print(next(gen))     # 4
print(next(gen))     # 9
```

```python
#closure example
def make_multiplier(factor):
    def multiply(n):
        return n * factor
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)

print(double(5))
print(triple(5))

```

```python
#Decorator example
def my_decorator(func):     #my decorator is the outer function and func is the original func used as input here
    def wrapper():        #Function inside function  which would be called before and after the final output
        print("Taransh is the goat")       #would be printed before the actual function
        func()   #original function called here wrapped
        print("This is after") #this is after 
    return wrapper #return the wrapper function 


def hello():   # this is the function that would be put in between
    print("Hello")

hello = my_decorator(hello)  #add the function with decorator 
hello()  #call it

# we can alro write it in a way - @my_decorator 
#def hello()
    #print("II")

    #hello()
  


```

```python
def make_adder(factor):
    def add(n):
        return n + factor
    return add

add5 = make_adder(10)

print(add5(2))
    
        
```

```python
def log_time(func):
    def huhu():
        print("Function started")
        func()
        print("Function ended")
    return huhu          # ✅ return it, don't call it

@log_time
def testing():
    print("I love python")

testing()
```
