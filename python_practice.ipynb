```python
#inheritance

class Animal:
    def __init__(self,name):
        self.name = name

    def speak(self):
        return  f"{self.name} says woof!"


class Dog(Animal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed = breed

    def fetch(self):
        return f"{self.name}  is of the breed {self.breed}"

d1 = Dog("sherry", "Rottweiler")


print(d1.name)
print(d1.fetch())
print(d1.speak())
```

```python
class Vehicle:  #made a parent class vehicle
    def __init__(self, brand, speed):  #initial function dunder method init whic python calls itself, self is a must rest two attributes can be specific
        self.speed = speed #always has to make sure atrributes are called with self
        self.brand = brand

    def is_fast(self):  #funtion to chekc if car is ffast enough
        return self.speed > 120  #speed would be passed and check should be more than 120

    def __str__(self):   #tells python what text represents this object, could use print but it just displays shit on screen
        return f"Brand: {self.brand} | Speed: {self.speed}"


class ElectricVehicle(Vehicle):
    def __init__(self, brand, speed, battery_range):
        super().__init__(brand, speed) #super is used to inherit methoda and attributes from other classes or parent class
        self.battery_range = battery_range

    def show_range(self):
        return f"{self.brand} has a range of {self.battery_range} km"

    def __str__(self):
        return super().__str__() + f" | Range: {self.battery_range} km"


v1 = Vehicle("Mahindra XUV", 200)
print(v1)
print(v1.is_fast())

e1 = ElectricVehicle("Skoda Taigun", 180, 550)
print(e1)
print(e1.is_fast())
print(e1.show_range())
```
