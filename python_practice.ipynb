```python
#inheritance

class Animal:
    def __init__(self,name):
        self.name = name

    def speak(self):
        return  f"{self.name} says woof!"


class Dog(Animal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed = breed

    def fetch(self):
        return f"{self.name}  is of the breed {self.breed}"

d1 = Dog("sherry", "Rottweiler")


print(d1.name)
print(d1.fetch())
print(d1.speak())
```

```python
class Vehicle:  #made a parent class vehicle
    def __init__(self, brand, speed):  #initial function dunder method init whic python calls itself, self is a must rest two attributes can be specific
        self.speed = speed #always has to make sure atrributes are called with self
        self.brand = brand

    def is_fast(self):  #funtion to chekc if car is ffast enough
        return self.speed > 120  #speed would be passed and check should be more than 120

    def __str__(self):   #tells python what text represents this object, could use print but it just displays shit on screen
        return f"Brand: {self.brand} | Speed: {self.speed}"


class ElectricVehicle(Vehicle):
    def __init__(self, brand, speed, battery_range):
        super().__init__(brand, speed) #super is used to inherit methoda and attributes from other classes or parent class
        self.battery_range = battery_range

    def show_range(self):
        return f"{self.brand} has a range of {self.battery_range} km"

    def __str__(self):
        return super().__str__() + f" | Range: {self.battery_range} km"


v1 = Vehicle("Mahindra XUV", 200)
print(v1)
print(v1.is_fast())

e1 = ElectricVehicle("Skoda Taigun", 180, 550)
print(e1)
print(e1.is_fast())
print(e1.show_range())
```

```python
# Iterators and Generators 

def even_numbers(n):
    for i in range(0,n+1):
        if i % 2 ==  0:
            yield i

for num in even_numbers(10):
    print(num)
```

```python
def squares(n):
    for square in range(1, n + 1):
        yield square ** 2

gen = squares(3)
print(next(gen))     # 1
print(next(gen))     # 4
print(next(gen))     # 9
```

```python
class Book:
    def __init__(self, title, author, available=True):
        self.title = title
        self.author = author
        self.available = available

    def __str__(self):
        return f"{self.title} by {self.author} (Available: {self.available})"


# Decorator class ke BAHAR, upar likhte hain
def log_action(func):
    def wrapper(*args, **kwargs):          # *args/**kwargs: borrow_book ko (self, title) chahiye
        print(f"Action: {func.__name__}")  # function ka naam, ugly object nahi
        return func(*args, **kwargs)       # asli function chalao aur uska result wapas do
    return wrapper                         # return wrapper, call nahi (wrapper())


class Library:
    def __init__(self, name):
        self.name = name
        self.books = []                    # har Library ki apni alag list

    def add_book(self, book):
        self.books.append(book)            # list mein Book object daalo

    def list_books(self):
        for book in self.books:
            print(book)                    # print() khud Book ka __str__ chala deta hai

    @log_action
    def borrow_book(self, title):
        for book in self.books:
            if book.title == title:        # book ka title, library ka nahi
                if book.available:
                    book.available = False # us book ka status badlo
                    return f"Borrowed: {title}"
                else:
                    raise ValueError("Book already borrowed")
        raise ValueError("Book not found") # loop ke BAHAR: poori list dekh li, kuch nahi mila

    def available_books(self):
        for book in self.books:
            if book.available:
                yield book.title           # append ki jagah yield


# Use karna
lib = Library("Abes")
lib.add_book(Book("AOT", "Isayama"))
lib.add_book(Book("JJK", "Akutami"))
lib.add_book(Book("Death Note", "Ohba"))

lib.list_books()

print(lib.borrow_book("AOT"))

try:
    lib.borrow_book("AOT")                 # dobara wahi book
except ValueError as e:
    print(e)

try:
    lib.borrow_book("Naruto")              # jo exist hi nahi karti
except ValueError as e:
    print(e)

for title in lib.available_books():
    print(title)
```

```python
def count_frequency(numbers):
    counter = {}   #empty dict
    for number in numbers:  #loops through numbers
        if number in counter:
            counter[number]+=1  ## this one just check if the number is already in it or not
        else:
            counter[number] = 1  #####important one ; dict is empty, this condition adds the numbers in it we explicitly dont have to add the numbers

    return counter

numbers = [1,1,1,2,3,4,4,5]
result = count_frequency(numbers)
print(result)

## We learned that we make a empty dict but doesnt add numbers subjectively, the operations doest it by itself
```

```python

class BankAccount:
    def __init__(self, holder, balance):
        if balance < 0:
            raise ValueError("Initial balance cannot be negative")

        self.holder = holder
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be greater than zero")

        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal must be greater than zero")

        if amount > self.balance:
            raise ValueError("Insufficient funds")

        self.balance -= amount
        return self.balance

    def check_balance(self):
        return self.balance


account = BankAccount("Bruce Wayne", 1000)

account.deposit(500)
account.withdraw(300)

print(account.check_balance())
```

```python
def second_largest(numbers):
    pointer = 0

    for number in numbers:
        if number[pointer] < number[pointer+1]:
            number[pointer+1] = number[pointer]
    return number[pointer] 
```
